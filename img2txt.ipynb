```python
###### main Stuctured code to save image into csv with respective fields" ######
```

```python


invoice_no = re.search(r"Invoice no:\s*(\d+)", text)
date = re.search(r"Date of issue:\s*([0-9/]+)", text)
total = re.search(r"Total \$\s*([0-9.,]+)", text)


data = {
    "invoice_no": invoice_no.group(1) if invoice_no else "",
    "date": date.group(1) if date else "",
    "total_amount": total.group(1) if total else ""
}

df = pd.DataFrame([data])
df.to_csv("tabular-field.csv", index=False)
```
